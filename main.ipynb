```python
import librosa
import numpy as np
import os

os.makedirs("cache/ravdess_mfcc", exist_ok=True)

def extract_mfcc(filepath, n_mfcc=40, max_len=200):
    y, sr = librosa.load(filepath, sr=16000)
    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=n_mfcc)
    if mfcc.shape[1] < max_len:
        pad_width = max_len - mfcc.shape[1]
        mfcc = np.pad(mfcc, ((0, 0), (0, pad_width)), mode="constant")
    else:
        mfcc = mfcc[:, :max_len]
    return mfcc.astype(np.float32)

for i, row in df.iterrows():
    fname = os.path.basename(row["filepath"])
    out_path = f"cache/ravdess_mfcc/{fname}.npy"
    if os.path.exists(out_path):
        continue
    mfcc = extract_mfcc(row["filepath"])
    np.save(out_path, mfcc)
    if i % 200 == 0:
        print(f"{i}/{len(df)} done")

print("MFCC extraction complete.")
print("Files in cache:", len(os.listdir("cache/ravdess_mfcc")))
```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score

LABELS = ["neutral","calm","happy","sad","angry","fearful","disgust","surprised"]
label2idx = {l: i for i, l in enumerate(LABELS)}

class RavdessMFCCDataset(Dataset):
    def __init__(self, df, split):
        self.df = df[df["split"] == split].reset_index(drop=True)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        fname = os.path.basename(row["filepath"])
        mfcc = np.load(f"cache/ravdess_mfcc/{fname}.npy")   # shape [40, 200]
        label = label2idx[row["emotion"]]
        return torch.tensor(mfcc), label

class SimpleMFCCClassifier(nn.Module):
    def __init__(self, n_mfcc=40, n_classes=8):
        super().__init__()
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(
            nn.Linear(n_mfcc, 128), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(128, n_classes)
        )

    def forward(self, x):              # x: [batch, 40, 200]
        x = self.pool(x).squeeze(-1)   # [batch, 40]
        return self.fc(x)

device = "cuda" if torch.cuda.is_available() else "cpu"

train_ds = RavdessMFCCDataset(df, "train")
val_ds   = RavdessMFCCDataset(df, "val")
train_loader = DataLoader(train_ds, batch_size=16, shuffle=True)
val_loader   = DataLoader(val_ds, batch_size=16)

model = SimpleMFCCClassifier().to(device)
opt = torch.optim.AdamW(model.parameters(), lr=1e-3)
loss_fn = nn.CrossEntropyLoss()

def evaluate(model, loader):
    model.eval()
    preds, trues = [], []
    with torch.no_grad():
        for x, y in loader:
            x = x.to(device)
            pred = model(x).argmax(dim=1).cpu()
            preds.extend(pred.tolist())
            trues.extend(y.tolist())
    return {
        "WA": accuracy_score(trues, preds),
        "UA": balanced_accuracy_score(trues, preds),
        "macro_f1": f1_score(trues, preds, average="macro"),
    }

for epoch in range(20):
    model.train()
    total_loss = 0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        opt.zero_grad()
        loss = loss_fn(model(x), y)
        loss.backward()
        opt.step()
        total_loss += loss.item()

    val_metrics = evaluate(model, val_loader)
    print(f"Epoch {epoch+1}: loss={total_loss/len(train_loader):.3f}  "
          f"val WA={val_metrics['WA']:.3f}  val UA={val_metrics['UA']:.3f}")

print("\nFinal validation metrics:", evaluate(model, val_loader))
```

```python
import torch
import torchaudio
from transformers import WavLMModel, Wav2Vec2FeatureExtractor
import os

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

extractor = Wav2Vec2FeatureExtractor.from_pretrained("microsoft/wavlm-base-plus")
wavlm_model = WavLMModel.from_pretrained("microsoft/wavlm-base-plus").to(device).eval()

def load_audio(path, target_sr=16000):
    wav, sr = torchaudio.load(path)
    if wav.shape[0] > 1:
        wav = wav.mean(dim=0, keepdim=True)
    if sr != target_sr:
        wav = torchaudio.functional.resample(wav, sr, target_sr)
    return wav.squeeze(0)

@torch.no_grad()
def extract_wavlm_embedding(path, n_tokens=16):
    wav = load_audio(path)
    inputs = extractor(wav.numpy(), sampling_rate=16000, return_tensors="pt")
    inputs = {k: v.to(device) for k, v in inputs.items()}
    out = wavlm_model(**inputs).last_hidden_state.squeeze(0)   

    out = out.unsqueeze(0).transpose(1, 2)                     
    out = torch.nn.functional.adaptive_avg_pool1d(out, n_tokens)
    out = out.transpose(1, 2).squeeze(0)                       
    return out.half().cpu()

os.makedirs("cache/ravdess_wavlm", exist_ok=True)

for i, row in df.iterrows():
    fname = os.path.basename(row["filepath"])
    out_path = f"cache/ravdess_wavlm/{fname}.pt"
    if os.path.exists(out_path):
        continue
    emb = extract_wavlm_embedding(row["filepath"])
    torch.save(emb, out_path)
    if i % 100 == 0:
        print(f"{i}/{len(df)} done")

print("WavLM extraction complete.")
print("Files in cache:", len(os.listdir("cache/ravdess_wavlm")))
```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score
import os

LABELS = ["neutral","calm","happy","sad","angry","fearful","disgust","surprised"]
label2idx = {l: i for i, l in enumerate(LABELS)}

class RavdessWavLMDataset(Dataset):
    def __init__(self, df, split):
        self.df = df[df["split"] == split].reset_index(drop=True)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        fname = os.path.basename(row["filepath"])
        emb = torch.load(f"cache/ravdess_wavlm/{fname}.pt").float()   # [16, 768]
        label = label2idx[row["emotion"]]
        return emb, label

class WavLMClassifier(nn.Module):
    def __init__(self, dim=768, n_classes=8):
        super().__init__()
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(
            nn.Linear(dim, 256), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(256, n_classes)
        )

    def forward(self, x):            # x: [batch, 16, 768]
        x = x.transpose(1, 2)        # [batch, 768, 16]
        x = self.pool(x).squeeze(-1) # [batch, 768]
        return self.fc(x)

device = "cuda" if torch.cuda.is_available() else "cpu"

train_ds = RavdessWavLMDataset(df, "train")
val_ds   = RavdessWavLMDataset(df, "val")
train_loader = DataLoader(train_ds, batch_size=16, shuffle=True)
val_loader   = DataLoader(val_ds, batch_size=16)

model = WavLMClassifier().to(device)
opt = torch.optim.AdamW(model.parameters(), lr=1e-4)
loss_fn = nn.CrossEntropyLoss()

def evaluate(model, loader):
    model.eval()
    preds, trues = [], []
    with torch.no_grad():
        for x, y in loader:
            x = x.to(device)
            pred = model(x).argmax(dim=1).cpu()
            preds.extend(pred.tolist())
            trues.extend(y.tolist())
    return {
        "WA": accuracy_score(trues, preds),
        "UA": balanced_accuracy_score(trues, preds),
        "macro_f1": f1_score(trues, preds, average="macro"),
    }

for epoch in range(20):
    model.train()
    total_loss = 0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        opt.zero_grad()
        loss = loss_fn(model(x), y)
        loss.backward()
        opt.step()
        total_loss += loss.item()

    val_metrics = evaluate(model, val_loader)
    print(f"Epoch {epoch+1}: loss={total_loss/len(train_loader):.3f}  "
          f"val WA={val_metrics['WA']:.3f}  val UA={val_metrics['UA']:.3f}")

print("\nFinal validation metrics:", evaluate(model, val_loader))
```

```python
class WavLMLayerWeightedClassifier(nn.Module):
    def __init__(self, n_layers=13, dim=768, n_classes=8):
        super().__init__()
        self.layer_weights = nn.Parameter(torch.zeros(n_layers))  # learned
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(
            nn.Linear(dim, 256), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(256, n_classes)
        )

    def forward(self, x):                      # x: [batch, layers, tokens, 768]
        w = torch.softmax(self.layer_weights, dim=0)   # [layers]
        x = (x * w.view(1, -1, 1, 1)).sum(dim=1)        # [batch, tokens, 768]
        x = x.transpose(1, 2)
        x = self.pool(x).squeeze(-1)
        return self.fc(x)
```

```python
best_ua = 0
patience = 8
epochs_without_improvement = 0

for epoch in range(60):
    model.train()
    total_loss = 0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        opt.zero_grad()
        loss = loss_fn(model(x), y)
        loss.backward()
        opt.step()
        total_loss += loss.item()

    val_metrics = evaluate(model, val_loader)
    print(f"Epoch {epoch+1}: loss={total_loss/len(train_loader):.3f}  "
          f"val WA={val_metrics['WA']:.3f}  val UA={val_metrics['UA']:.3f}")

    if val_metrics["UA"] > best_ua:
        best_ua = val_metrics["UA"]
        epochs_without_improvement = 0
        torch.save(model.state_dict(), "results/best_ravdess_wavlm.pt")
    else:
        epochs_without_improvement += 1
        if epochs_without_improvement >= patience:
            print(f"Early stopping at epoch {epoch+1}")
            break

print("Best val UA:", best_ua)
```
